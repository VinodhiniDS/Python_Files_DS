## if, elif, else Use-case:

### 1. Positive, negative, or zero. Ask for a number and print whether it’s positive, negative, or zero.

```python
x= int(input("Enter the value: "))
if x>0:
    print("The value is positive:",x)
elif x<0:
    print("The value is negative:",x)
else:
    print("The value is Zero:",x)
```

### 2.Grade calculator.
Ask for a score (0-100) and print a letter grade: 90+ is A, 80-89 is B, 70-79 is C, 60-69 is D, and below 60 is F. If the score is outside 0-100, print “Invalid score”.

```python
score=float(input("Enter your mark: "))
if score <=0 or score>100:
    print("Invalid Score")
elif score >=90:
    print("Grade A",score)
elif score>=80:
    print("Grade B",score)
elif score >=70:
    print("Grade C",score)
elif score >=60:
    print("Grade D",score)
else:
    print("Grade F",score)
```

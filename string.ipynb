##Python String Use cases:

###1. Clean column names (strip, lower, replace)
Turn messy headers into snake_case.
[" Customer ID ", "Total Spend ($)", "Signup-Date"] → ["customer_id", "total_spend", "signup_date"]

```python
#using pandas index

import pandas as pd

column=[" Customer ID ", "Total Spend ($)", "Signup-Date"]

col=(pd.Index(column).str.lower().str.strip().str.replace(r"[^A-Za-z0-9]+","_",regex=True ).str.strip("-"))
col
```

###2. Extract email domains (split, partition, Counter)
Given a list of emails, return the domain of each and count how often each domain appears.
["ana@gmail.com", "raj@acme.io", "li@gmail.com"] → {"gmail.com": 2, "acme.io": 1}

```python
from collections import Counter

emails = ["ana@gmail.com", "raj@acme.io", "li@gmail.com"]

domains = [e.rpartition("@")[2].lower() for e in emails]   # domain of each email
counts = dict(Counter(domains))                            # frequency of each domain

print(domains)  # ['gmail.com', 'acme.io', 'gmail.com']
print(counts)   # {'gmail.com': 2, 'acme.io': 1}
```

###4. Split an email address (user accounts)
From "uma.sharma@example.com", extract the username and the domain separately.
Expected: uma.sharma and example.com
Skills: split(), unpacking

```python
username="uma.sharma@example.com"

email=username.split("@")     #only split() method
email
```

```python
email="uma.sharma@example.com"

domain, username=email.split("@")  #unpacking and split() method
print(domain)
print(username)
```

###5. Count vowels (text analysis)
Count the vowels in "Absolutely fantastic service".
Expected: 10
Skills: loops, in, lower()

```python
vowels="Absolutely fantastic service thank you"

count=0
for i in vowels.lower():
    if i in "aeiou":
        count+=1
print(count)
```
